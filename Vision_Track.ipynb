```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
results  = model("/content/16849658_1080_1920_30fps.mp4" , save = True)
```

```python
r = results[0]
print(len(results))
print(r.names)
print(r.boxes.xyxy)
print(r.boxes.conf)
print(r.boxes.cls)
```

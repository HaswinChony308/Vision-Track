### Step 1: Download a Sample Video
Let's download a sample video from a public URL to `/content/16849658_1080_1920_30fps.mp4` so that your object detection and tracking code can run successfully.

```python
import urllib.request
import os

# Using a highly reliable, direct public sample video source from Github
video_url = "https://raw.githubusercontent.com/intel-iot-devkit/sample-videos/master/people-detection.mp4"
output_path = "/content/16849658_1080_1920_30fps.mp4"

print("Downloading stable sample video...")
req = urllib.request.Request(
    video_url,
    headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
)

with urllib.request.urlopen(req) as response, open(output_path, 'wb') as out_file:
    out_file.write(response.read())

print(f"Video downloaded successfully and saved to: {output_path}")
```

### Step 2: Object Tracking and Counting Unique Objects
Now that the video is downloaded, we can initialize the model, track objects across frames, and count how many **unique** people (class `0`) were seen using `results.boxes.id` and Python sets.

```python
from ultralytics import YOLO

# Load the YOLOv8 model
model = YOLO("yolov8n.pt")

# Run tracking on the downloaded video
results = model.track("/content/16849658_1080_1920_30fps.mp4", tracker="bytetrack.yaml", persist=True)

# Set to keep track of unique person IDs
unique_person_ids = set()

for r in results:
    # Check if boxes and track IDs exist in the frame
    if r.boxes is not None and r.boxes.id is not None:
        classes = r.boxes.cls.tolist()
        track_ids = r.boxes.id.tolist()

        # Iterate through detected objects
        for cls, track_id in zip(classes, track_ids):
            # Check if the class is a person (class 0)
            if int(cls) == 0:
                unique_person_ids.add(int(track_id))

print(f"\nDetected Tracked IDs for people: {unique_person_ids}")
print(f"Total unique people detected throughout the video: {len(unique_person_ids)}")
```

** 1 Fix Person Counting & Collect Timeline **

```python
!pip install -q ultralytics
import os
import urllib.request
from ultralytics import YOLO

# Ensure the video is downloaded
video_path = "/content/16849658_1080_1920_30fps.mp4"
if not os.path.exists(video_path):
    print("Downloading sample video...")
    video_url = "https://raw.githubusercontent.com/intel-iot-devkit/sample-videos/master/people-detection.mp4"
    req = urllib.request.Request(video_url, headers={'User-Agent': 'Mozilla/5.0'})
    with urllib.request.urlopen(req) as response, open(video_path, 'wb') as out_file:
        out_file.write(response.read())
    print("Video downloaded successfully!")

# Initialize model and ensure results are populated
model = YOLO("yolov8n.pt")
results = model.track(video_path, tracker="bytetrack.yaml", persist=True)

# Extract person counts (class 0) per frame
person_counts = []

for r in results:
    classes = r.boxes.cls.tolist()
    # Count occurrences of class 0 ('person')
    count = classes.count(0)
    person_counts.append(count)

print(f"Total frames processed: {len(person_counts)}")
print(f"Max people in a single frame: {max(person_counts)}")
print(f"Average people per frame: {sum(person_counts) / len(person_counts):.2f}")
```

2. Ploting People Count Over Time (Video Duration)

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(person_counts, color='royalblue', linewidth=1.5, label='Detected Persons')
plt.title('Person Count per Frame Over Time')
plt.xlabel('Frame Number')
plt.ylabel('Number of People Detected')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

3. Move from Pure Detection to Object Tracking (model.track)
Right now you are doing frame-by-frame independent detection. You cannot tell if person #1 in frame 1 is the same person in frame 2, nor can you count the total unique individuals that passed through the camera view.

To assign persistent IDs across frames, switch from model(...) to model.track(...):

```python
!pip install -q ultralytics
import os
import urllib.request
from ultralytics import YOLO

# Ensure the video is downloaded
video_path = "/content/16849658_1080_1920_30fps.mp4"
if not os.path.exists(video_path):
    print("Downloading sample video...")
    video_url = "https://raw.githubusercontent.com/intel-iot-devkit/sample-videos/master/people-detection.mp4"
    req = urllib.request.Request(video_url, headers={'User-Agent': 'Mozilla/5.0'})
    with urllib.request.urlopen(req) as response, open(video_path, 'wb') as out_file:
        out_file.write(response.read())
    print("Video downloaded successfully!")

# Initialize model
model = YOLO("yolov8n.pt")

# Track objects across frames using ByteTRACK or BoT-SORT
tracking_results = model.track(
    source=video_path,
    tracker="bytetrack.yaml",
    stream=True,
    classes=[0],
    save=False
)

unique_ids = set()

for frame_idx, r in enumerate(tracking_results):
    if r.boxes.id is not None:
        track_ids = r.boxes.id.int().tolist()
        unique_ids.update(track_ids)

print(f"Total unique people tracked across the video: {len(unique_ids)}")
```

### Step 2: Object Tracking and Counting Unique Objects
Now that the video is downloaded, we can initialize the model, track objects across frames, and count how many **unique** people (class `0`) were seen using `results.boxes.id` and Python sets.

```python
from ultralytics import YOLO

# Load the YOLOv8 model
model = YOLO("yolov8n.pt")

# Run tracking on the downloaded video
results = model.track("/content/16849658_1080_1920_30fps.mp4", tracker="bytetrack.yaml", persist=True)

# Set to keep track of unique person IDs
unique_person_ids = set()

for r in results:
    # Check if boxes and track IDs exist in the frame
    if r.boxes is not None and r.boxes.id is not None:
        classes = r.boxes.cls.tolist()
        track_ids = r.boxes.id.tolist()

        # Iterate through detected objects
        for cls, track_id in zip(classes, track_ids):
            # Check if the class is a person (class 0)
            if int(cls) == 0:
                unique_person_ids.add(int(track_id))

print(f"\nDetected Tracked IDs for people: {unique_person_ids}")
print(f"Total unique people detected throughout the video: {len(unique_person_ids)}")
```

**Region of Interest (ROI) Filtering / Line Crossing Count**
If you only want to count objects that pass through a specific virtual zone or coordinate boundary (useful for traffic monitoring or entrance counters), you can check if the object's center point falls within a defined bounding box:

```python
# Define your Region of Interest (ROI) as [xmin, ymin, xmax, ymax]
# Adjust these pixel coordinates based on your video dimensions
roi = [200, 400, 800, 900]

roi_counts = []

for r in results:
    if r.boxes is not None and r.boxes.xyxy is not None:
        boxes = r.boxes.xyxy.cpu().numpy()
        classes = r.boxes.cls.cpu().numpy()

        frame_roi_count = 0
        for box, cls in zip(boxes, classes):
            if int(cls) == 0: # Person
                x1, y1, x2, y2 = box
                # Calculate center point of the bounding box
                center_x = (x1 + x2) / 2
                center_y = (y1 + y2) / 2

                # Check if center falls inside ROI
                if roi[0] <= center_x <= roi[2] and roi[1] <= center_y <= roi[3]:
                    frame_roi_count += 1

        roi_counts.append(frame_roi_count)
    else:
        roi_counts.append(0)

print(f"Max people inside ROI in a single frame: {max(roi_counts)}")
```

### Feature 5: Pedestrian Spatial Heatmap (Traffic Density Analysis)
This feature aggregates the center point coordinates of all detected people across every frame and overlays a density heatmap. This is highly useful for retail, security, and urban planning to identify high-dwell zones or popular walking paths.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Initialize YOLO model and open video
model = YOLO("yolov8n.pt")
video_path = "/content/16849658_1080_1920_30fps.mp4"
cap = cv2.VideoCapture(video_path)

# Get video dimensions to build our accumulator map
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Create a blank 2D accumulator grid
heatmap_accumulator = np.zeros((height, width), dtype=np.float32)

print("Processing frames to build spatial heatmap...")
frame_idx = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Process every 3rd frame to optimize speed
    if frame_idx % 3 == 0:
        results = model(frame, verbose=False)
        r = results[0]
        if r.boxes is not None:
            boxes = r.boxes.xyxy.cpu().numpy()
            classes = r.boxes.cls.cpu().numpy()

            for box, cls in zip(boxes, classes):
                if int(cls) == 0:  # Class 0 represents 'person'
                    x1, y1, x2, y2 = map(int, box)
                    # Calculate center point
                    cx, cy = (x1 + x2) // 2, (y1 + y2) // 2

                    # Draw a localized gaussian-like intensity point on our grid
                    # Ensure coordinates are safely within image boundaries
                    if 0 <= cx < width and 0 <= cy < height:
                        cv2.circle(heatmap_accumulator, (cx, cy), 15, 1.0, -1)

    frame_idx += 1

cap.release()

# Apply a Gaussian blur to smooth the discrete points into a continuous heat distribution
heatmap_blurred = cv2.GaussianBlur(heatmap_accumulator, (51, 51), 0)

# Normalize the heatmap between 0 and 255
if np.max(heatmap_blurred) > 0:
    heatmap_normalized = (heatmap_blurred / np.max(heatmap_blurred) * 255).astype(np.uint8)
else:
    heatmap_normalized = heatmap_blurred.astype(np.uint8)

# Plot the heatmap
plt.figure(figsize=(12, 7))
plt.imshow(heatmap_normalized, cmap='jet')
plt.title("Pedestrian Traffic Density Spatial Heatmap")
plt.xlabel("X Coordinate (Pixels)")
plt.ylabel("Y Coordinate (Pixels)")
plt.colorbar(label="Traffic Density Intensity")
plt.show()
```

### Feature 6: Directional Flow Counter (Left-to-Right vs. Right-to-Left)
This feature uses object tracking to monitor individual trajectories and count how many people cross a central threshold moving left-to-right versus right-to-left.

```python
from ultralytics import YOLO
import cv2

model = YOLO("yolov8n.pt")
video_path = "/content/16849658_1080_1920_30fps.mp4"

# Track movement trajectories
trajectories = {}

# Direction counts
left_to_right = 0
right_to_left = 0

# Define a vertical midline split in the video (e.g., center X pixel coordinate)
cap = cv2.VideoCapture(video_path)
midline_x = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)) // 2
cap.release()

print(f"Analyzing directional flow relative to midline X = {midline_x}...")

# Run tracking stream
tracking_stream = model.track(
    source=video_path,
    tracker="bytetrack.yaml",
    stream=True,
    classes=[0],  # only track people
    persist=True
)

for frame_idx, r in enumerate(tracking_stream):
    if r.boxes is not None and r.boxes.id is not None:
        boxes = r.boxes.xyxy.cpu().numpy()
        track_ids = r.boxes.id.int().tolist()

        for box, track_id in zip(boxes, track_ids):
            x1, y1, x2, y2 = box
            cx = (x1 + x2) / 2

            if track_id not in trajectories:
                trajectories[track_id] = [cx]
            else:
                trajectories[track_id].append(cx)

# Classify trajectories based on their transition across the midline
processed_ids = set()
for track_id, x_coords in trajectories.items():
    if len(x_coords) > 5:  # Ensure enough historical data is present
        start_x = x_coords[0]
        end_x = x_coords[-1]

        # Check if the path crossed the central threshold
        if start_x < midline_x and end_x > midline_x:
            left_to_right += 1
        elif start_x > midline_x and end_x < midline_x:
            right_to_left += 1

print(f"\n--- Directional Flow Analysis ---")
print(f"Left-to-Right crossings: {left_to_right}")
print(f"Right-to-Left crossings: {right_to_left}")
print(f"Total unique people tracked: {len(trajectories)}")
```

### Feature 7: Pedestrian Dwell Time Estimation
Using the persistent tracking data, we can estimate how long (in seconds) each person spends in the camera's field of view. This is calculated by dividing the number of frames a specific ID was detected by the video's frame rate (30 FPS).

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming 30 FPS for our video source
FPS = 30.0

dwell_times = {}
for track_id, coordinates in trajectories.items():
    frame_count = len(coordinates)
    dwell_seconds = frame_count / FPS
    dwell_times[track_id] = dwell_seconds

# Convert to DataFrame
df_dwell = pd.DataFrame(list(dwell_times.items()), columns=['Track ID', 'Dwell Time (seconds)'])

print("--- Pedestrian Dwell Times ---")
print(df_dwell.to_string(index=False))

# Plot the dwell times
plt.figure(figsize=(10, 5))
plt.bar(df_dwell['Track ID'].astype(str), df_dwell['Dwell Time (seconds)'], color='orange', edgecolor='black')
plt.title('Dwell Time per Tracked Individual')
plt.xlabel('Person ID')
plt.ylabel('Duration in Frame (Seconds)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Feature 9: Class-Specific Multi-Object Tracking & Dashboard
This feature extends the tracking logic to monitor multiple distinct classes simultaneously (e.g., people, cars, and bicycles) and provides a visual dashboard of unique counts across all target classes.

```python
!pip install -q ultralytics
import matplotlib.pyplot as plt
import pandas as pd
from ultralytics import YOLO

# Initialize model
model = YOLO("yolov8n.pt")
video_path = "/content/16849658_1080_1920_30fps.mp4"

# Map COCO Class IDs we want to track
# 0: Person, 1: Bicycle, 2: Car
target_classes = {0: "Person", 1: "Bicycle", 2: "Car"}
unique_counts = {name: set() for name in target_classes.values()}

print("Running multi-class tracking...")
tracking_stream = model.track(
    source=video_path,
    tracker="bytetrack.yaml",
    stream=True,
    classes=list(target_classes.keys()),
    persist=True,
    verbose=False
)

for r in tracking_stream:
    if r.boxes is not None and r.boxes.id is not None:
        classes = r.boxes.cls.int().tolist()
        track_ids = r.boxes.id.int().tolist()

        for cls, track_id in zip(classes, track_ids):
            if cls in target_classes:
                class_name = target_classes[cls]
                unique_counts[class_name].add(track_id)

# Prepare dashboard data
dashboard_data = {class_name: len(ids) for class_name, ids in unique_counts.items()}
print("\n--- Multi-Object Tracking Summary ---")
for class_name, count in dashboard_data.items():
    print(f"Unique {class_name}s Detected: {count}")

# Plot Dashboard
plt.figure(figsize=(8, 5))
plt.bar(dashboard_data.keys(), dashboard_data.values(), color=['salmon', 'lightgreen', 'skyblue'], edgecolor='black')
plt.title('Dashboard: Total Unique Tracked Objects by Class')
plt.ylabel('Unique Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Feature 10: Speed Benchmarking & FPS Profiler
This feature analyzes the computational latency of each processing step (Preprocessing, Inference, and Postprocessing) across all processed video frames and calculates the average Processing FPS to benchmark performance.

```python
!pip install -q ultralytics
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
video_path = "/content/16849658_1080_1920_30fps.mp4"

preprocess_times = []
inference_times = []
postprocess_times = []

print("Profiling model performance...")
results = model(video_path, verbose=False)

for r in results:
    if r.speed is not None:
        preprocess_times.append(r.speed.get('preprocess', 0.0))
        inference_times.append(r.speed.get('inference', 0.0))
        postprocess_times.append(r.speed.get('postprocess', 0.0))

# Calculate averages in milliseconds
avg_prep = np.mean(preprocess_times)
avg_inf = np.mean(inference_times)
avg_post = np.mean(postprocess_times)
total_latency = avg_prep + avg_inf + avg_post

# Calculate estimated FPS (1000ms / latency in ms)
est_fps = 1000.0 / total_latency if total_latency > 0 else 0.0

print("\n--- Performance Profiling Results ---")
print(f"Average Preprocess Time: {avg_prep:.2f} ms")
print(f"Average Inference Time:  {avg_inf:.2f} ms")
print(f"Average Postprocess Time: {avg_post:.2f} ms")
print(f"Total End-to-End Latency: {total_latency:.2f} ms per frame")
print(f"Estimated Processing Speed: {est_fps:.2f} FPS")

# Visualize latency breakdown
steps = ['Preprocess', 'Inference', 'Postprocess']
times = [avg_prep, avg_inf, avg_post]

plt.figure(figsize=(8, 5))
plt.bar(steps, times, color=['orchid', 'goldenrod', 'tomato'], edgecolor='black')
plt.title('Computational Latency Breakdown (Lower is Better)')
plt.ylabel('Time (Milliseconds)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
